```python
import random
class Spliterator:
    def __init__(self, rate = 0.2, shuffle = True) -> None:
        self.rate = rate
        self.shuffle = shuffle
    def fit(self, data, random_state = 0):
        if self.shuffle:
            if random_state:
                random.seed(random_state)
            random.shuffle(data) 
            shuffle_data = data 
        else:
            shuffle_data = data 
        ind_split = int(self.rate*len(data))
        train = shuffle_data[:-ind_split] 
        test = shuffle_data[-ind_split:]
        return train, test 
```

```python
data  =list(range(10))
split_model = Spliterator()
print(split_model.fit(data))
```

```python

class Scaler:
    def __init__(self, a=0, b=1) -> None:
        self.a = a
        self.b =b 
    def fit(self, data):
        max_value = max(data)
        min_value = min(data)
        scaled_data = list(map(lambda x: self.a+(self.b-self.a)*(x-min_value)/(max_value-min_value),
                               data))
        return scaled_data
```

```python
data = range(0,11)
scaler = Scaler(-1,1)
print(scaler.fit(data))
```

```python
# Stack
class Stack:
    def __init__(self):
        self.__stack = []
    def push(self,item):
        return self.__stack.append(item)
    def peek(self):
        if self.__stack:
            return self.__stack[-1]
        else:
            return "the stack is empty"
    def pop(self):
        if self.__stack:
            return self.__stack.pop()
        else:
            return "the stack is empty"
    def is_empty(self):
        return not bool(self.__stack)
    def size(self):
        return len(self.__stack)
    def __str__(self):
        return str(self.__stack)
    def __add__(self, other_stack):
        stack = list()
        return self.__stack+other_stack.__stack + self.__stack
    def __getitem__(self, key):
        return self.__stack[key]
```

```python
my_stack = Stack()
my_stack.push('item1')
my_stack.push('item2')
my_stack.push('item3')
my_stack.push('item4')

other_stack  = Stack()
other_stack.push('oitem1')
other_stack.push('oitem2')
other_stack.push('oitem3')
my_stack + other_stack
```

```python
class Email:
    existing_emails = []
    def __init__(self, email) -> None:
        if email not in Email.existing_emails:
            self.email = email
            self.inbox = []
            self.sent = []
            Email.existing_emails.append(email)
        else:
            print('your email is existed, pleas choose another!')
    def send(self,other, message):
        self.sent.append(message)
        other.inbox.append(message)
    def read(self):
        print(self.inbox)


    
```
